```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
import glob
```

```python
def leer_archivo(nombre_dataframe):
    if nombre_dataframe.endswith('.csv'):
        data = pd.read_csv(nombre_dataframe)
    elif nombre_dataframe.endswith('.xlsx'):
        data = pd.read_xlsx(nombre_dataframe)
    elif nombre_dataframe.endswith('.html'):
        data = pd.read_html(nombre_dataframe)
    elif nombre_dataframe.endswith('.json'):
        with open(nombre_dataframe) as f:
            data = json.load(f)
    else:
        print('Ingrese un archivo tipo csv, xlsx, html o json')
        
    return(data)
```

```python
leer_archivo('Copia de 001.json')
```
